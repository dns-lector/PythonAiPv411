|Початкове слово|Лемма|Стем|
|-|-|-|

**Лематизація** - процес групування різних форм слова з метою аналізу їх як єдиних за змістом (будинку, будинки -- лема: будинок)

**Стемінг** - процес скорочення слова до основи шляхом відкидання допоміжних частин, таких як закінчення чи суфіксу (будинку, будинки, будинок -- стем: будинк)

```python
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer, PorterStemmer 
lemmatizer = WordNetLemmatizer()
stemmer = PorterStemmer()
```

```python
# Перетворіть усі токени на леми і на стеми. Виведіть у вигляді таблиці.
from IPython.display import Markdown as md
md( "\n".join((
    "|Початкове слово|Лемма|Стем|",
    "|-|-|-|",
    *(f"|{w}|{lemmatizer.lemmatize(w)}|{stemmer.stem(w)}|" for w in feedbacks_tokens["positive"][:10] )
)) )
# *(1,2,3) -> 1,2,3
```

## Завдання 3
Видаліть стоп-слова і проаналізуйте частотність.
- Завантажте список англійських стоп-слів nltk.corpus.stopwords.words('english').
- Видаліть із токенів усі стоп-слова.
- Підрахуйте частоту слів, що залишилися, за допомогою collections.Counter.
- Виведіть 10 найчастіших слів та їхню кількість.

stopwords - слова, що не несуть змістового навантаження - сполучники, артиклі тощо.
Їх роль - синтаксична, але на тон фрази вони не впливають. Традиційно такі слова видаляють з загального масиву токенів перед аналізом.

```python
nltk.download('stopwords')
from nltk.corpus import stopwords
stop_words = stopwords.words('english')
print(len(stop_words))
print(stop_words[:10])
```

```python
# Видаліть із токенів усі стоп-слова
feedbacks_stop = {
    "positive": [ token  for token in feedbacks_tokens["positive"] if not token in stop_words ],
    "negative": [ token  for token in feedbacks_tokens["negative"] if not token in stop_words ],
}
print(len(feedbacks_tokens["positive"]), len(feedbacks_tokens["negative"]) )
print(len(feedbacks_stop['positive']), len(feedbacks_stop['negative']))
```

```python
# Підрахуйте частоту слів, що залишилися, за допомогою collections.Counter
from collections import Counter
feedbacks_freqs = {
    "positive": Counter( feedbacks_stop['positive'] ),
    "negative": Counter( feedbacks_stop['negative'] ),
}
# Виведіть 10 найчастіших слів та їхню кількість.
md( "\n".join((
    "|Cлово(+)|Кількість|Cлово(-)|Кількість|",
    "|-|-|-|-|",
    *(f"|{pos[0]}|{pos[1]}|{neg[0]}|{neg[1]}|"  
      for pos, neg in zip(
          sorted(
              list(feedbacks_freqs['positive'].items()), 
              key=lambda x: x[1], 
              reverse=True
          )[:10],
          sorted(
              list(feedbacks_freqs['negative'].items()), 
              key=lambda x: x[1], 
              reverse=True
          )[:10]
      )
     )
)))
```
