```python
# zip(arr1,arr2) - поєднує множини за принципом [ [arr1[0], arr2[0]], [arr1[1], arr2[1]], ... ] 
# map(fun, arr) - застосовує функцію fun до кожного елементу arr
# " ".join(arr) - поєднує всі елементи arr через пробіл (вимагає, щоб елементи arr були рядками)
print(
    *( f"{x} | {y}" for x, y in zip( arr2, np.sum(arr2, axis=1) ) ), 
    "------------------",
    # f"{np.sum(arr2, axis=0)} {np.mean(arr2)}",
    " ".join(map(str, (np.sum(arr2, axis=0), '|', np.mean(arr2)))),
    sep='\n'
)
```

## Завдання 4
Створіть масив розміром 10×10, заповнений випадковими числами з плаваючою крапкою від 0 до 1.
- Знайдіть мінімальне та максимальне значення в усьому масиві.
- Визначте індекси мінімального та максимального елементів
- (наприклад: (рядок, стовпець)).
- Виведіть сам масив і окремо знайдені значення разом з їхніми індексами.

```python
arr4 = np.random.rand(10,10)
# print(*(f"{x:.3f}" for a in arr4 for x in a)) - всі числа в один рядок
# ' '.join(f"{x:.3f}" for x in arr4[0]) - числа першого рядка масива
print(*(
        ' '.join(f"{x:.2f}" for x in a) 
                for a in arr4
      ), 
      sep='\n')  # всі рядки через \n
```

```python
# Визначте індекси мінімального та максимального елементів
min_index = np.argmin(arr4)  # індекс загальний, наче масив лінійний
min_index
```

```python
xy = np.unravel_index(min_index, arr4.shape)  # індекс перерахований на форму масиву (подвійний)
print("мінімум знаходиться за індексами:", ",".join( map(str, xy) ))
```

```python
print(
    "максимум знаходиться за індексами:", 
    ",".join( map(str, np.unravel_index(np.argmax(arr4), arr4.shape)) )
)
```

## Завдання 5
Створіть одновимірний масив із 15 випадкових цілих чисел від -10 до 10.
- Виведіть масив.
- За допомогою булевої індексації виберіть тільки додатні елементи.
- За допомогою булевої індексації виберіть тільки парні елементи.
- Виведіть обидва результати на екран.

```python
arr5 = np.random.randint(-10,10,size=15)
print(arr5)
arr5
```

```python
cond_index = arr5 > 0  # умова формує масив результатів для кожного з елементів
print(cond_index)
print(arr5[cond_index])  # умовна індексація
```

```python
print(arr5[arr5 % 2 == 0])  # умовна індексація одразу з умовою
```

```python
np.count_nonzero(arr5 % 2 == 0)  # підрахунок кількості елементів, що відповідають умові
```
