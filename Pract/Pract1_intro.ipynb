Вам необхідно підготувати в Jupyter Notebook короткий звіт, де є вступні дані, розрахунки та оформлені висновки.
- Створіть Markdown-комірку із заголовком "Звіт із продажів" і підпишіть: "У цьому розділі ми рахуємо підсумки за простими даними".
- У кодовій комірці створіть список продажів за тиждень:
sales = [120, 135, 100, 150, 90, 200, 170]
- Порахуйте:
    - загальна кількість продажів за тиждень;
    - середнє значення;
    - максимальне і мінімальне значення.
    - Створіть Markdown-комірку й оформіть таблицю з результатами:

|Показник|Значення|
|-|-|
|Загальна сума|...|
|Середнє|...|
|Максимум|...|
|Мінімум|...|

```python
sales = [120, 135, 100, 150, 90, 200, 170]
```

```python
from IPython.display import Markdown as md
# Спосіб 1 - використання інтерпольованих рядків
md(f"""|Показник|Значення|
|-|-|
|Загальна сума|{sum(sales)}|
|Середнє      |{sum(sales)/len(sales):.2f}|
|Максимум     |{max(sales)}|
|Мінімум      |{min(sales)}|""")
```

```python
# Спосіб 2 - кортеж підстановки
md("""|Показник|Значення|
|-|-|
|Загальна сума|%d|
|Середнє    |%.2f|
|Максимум   |%d|
|Мінімум    |%d|""" % (
    sum(sales), 
    sum(sales)/len(sales), 
    max(sales), 
    min(sales))
)
```

```python
# Спосіб 3 - HTML
md('''<table style="width:100%%"><tr><th>Показник</th><th>Значення</th></tr>
<tr><td>Загальна сума</td><td>%d  </td></tr>
<tr><td>Середнє      </td><td>%.2f</td></tr>
<tr><td>Максимум     </td><td>%d  </td></tr>
<tr><td>Мінімум      </td><td>%d  </td></tr>
</table>''' % (
    sum(sales), 
    sum(sales)/len(sales), 
    max(sales), 
    min(sales))
)
```

students = [("Alex", 85), ("Maria", 92), ("John", 74), ("Olga", 60), ("Max", 48)]

- Створіть "текстову візуалізацію": для кожного студента виведіть рядок Ім'я | *****, де кількість * залежить від десятків в оцінці (приклад: 92 → 9 зірок).
- У Markdown-комірці оформіть таблицю з результатами:

```python
students = [("Alex", 85), ("Maria", 92), ("John", 74), ("Olga", 60), ("Max", 48)]
```

```python
marks = [s[1] for s in students]
print( "Cередній бал", sum(marks) / len(students) )
max(marks)
```

```python
[(s[0], '*' * (s[1] // 10)) for s in students]
```
