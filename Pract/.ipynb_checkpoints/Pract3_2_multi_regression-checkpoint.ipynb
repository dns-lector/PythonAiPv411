# Множинна регресія та попереднє опрацювання даних
Множинна регресія дозволяє знаходити залежності від кількох параметрів.
Більш того, дозволяє провести оцінку внеску цих параметрів у результат.

## Завдання 1
Побудуйте модель лінійної регресії на кількох ознаках датасету California Housing:
- Завантажте датасет зі scikit-learn.
- Сформуйте DataFrame з ознаками та цільовою змінною.
- Виберіть 3 ознаки, наприклад: MedInc, HouseAge, AveRooms.
- Розділіть дані на навчальну і тестову вибірки (80/20, random_state=42).
- Навчіть модель LinearRegression і отримайте передбачення.
- Обчисліть R² на тестовій вибірці та виведіть результат у таблиці.

```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
```

```python
california_housing = fetch_california_housing(as_frame=True)
california_housing.frame.head()
```

```python
print(california_housing.DESCR)
```

```python
# Виберіть 3 ознаки, наприклад: MedInc, HouseAge, AveRooms.
housing = california_housing.frame[['MedInc', 'HouseAge', 'AveRooms']]
housing.head()
```

```python
# Розділіть дані на навчальну і тестову вибірки (80/20, random_state=42).
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(housing, california_housing.target, test_size=0.2, random_state=42)

print(f"Train: {len(ytrain)}, test: {len(ytest)}")
```

```python
# Навчіть модель LinearRegression і отримайте передбачення.
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(xtrain, ytrain)
```

```python
pred = model.predict(xtest)
print(*(f"{p:.2f} -- {y:.2f}" for p,y in zip(pred[:10], ytest[:10])), sep='\n')
```

```python
# Обчисліть R² на тестовій вибірці та виведіть результат у таблиці.
import sklearn.metrics as sm
r2_base = sm.r2_score(ytest, pred)
print("Коефіцієнт детермінації моделі", r2_base)
```

## Завдання 3
Застосуйте різні методи попередньої обробки даних:
- Додайте в DataFrame штучну категоріальну ознаку, наприклад:
import numpy as np 
df['Category'] = np.where(df['HouseAge'] > 30, 'Old', 'New')

- Перетворіть цю ознаку:
    - за допомогою factorize;
    - за допомогою OneHotEncoder.
- Додайте штучні пропуски в 1 колонку (наприклад, MedInc), заповніть їх через SimpleImputer (середнім значенням).
- Навчіть модель лінійної регресії та порівняйте R² для різних способів кодування категоріальної ознаки.

```python
import numpy as np 
housing['Category'] = np.where(housing['HouseAge'] > 30, 'Old', 'New')
housing.head()
```

```python
housing["factorize"] = pd.factorize(housing['Category'])[0]
housing.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(sparse_output=False, drop='first')  # sparse_output - формувати матрицю чи масив, drop - видалення додаткових варіантів
housing["encoder"] = encoder.fit_transform(housing['Category'].array.reshape(-1, 1))
housing.head()
```

```python
def score_model(data) :
    model = LinearRegression()
    xtrain, xtest, ytrain, ytest = train_test_split(data, california_housing.target, test_size=0.2, random_state=42)
    model.fit(xtrain, ytrain)
    return sm.r2_score(ytest, model.predict(xtest)), model
```

```python
score1, model1 = score_model(housing[['MedInc', 'HouseAge', 'AveRooms', 'factorize']])
print(score1)
model1
```

```python
score1, model1 = score_model(housing[['MedInc', 'HouseAge', 'AveRooms', 'encoder']])
print(score1)
model1
```

```python
# Додайте штучні пропуски в 1 колонку (наприклад, MedInc), ...
ind = [10,20,30,40,50]
housing.loc[ind, "MedInc"] = None
housing.loc[ind, "MedInc"]
```

```python
# ... заповніть їх через SimpleImputer (середнім значенням).
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy="mean")
housing["MedInc"] = imputer.fit_transform(housing["MedInc"].array.reshape(-1,1))
housing.loc[ind, "MedInc"]
```

```python
california_housing.frame.loc[ind, "MedInc"]
```

```python
score1, model1 = score_model(housing[['MedInc', 'HouseAge', 'AveRooms', 'encoder']])
print(score1)
model1
```
